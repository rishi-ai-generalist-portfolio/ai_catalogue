```python
from google import genai
from google.colab import userdata
from google.genai import types
from pydantic import BaseModel

client = genai.Client(api_key = userdata.get("GEMINI_API_KEY") )
csvfile = client.files.upload(file="DeliveryAgentFeedback.csv")
myprompt = "I am sending a csv file of Delivery Agents Feedback. I want \
to get the Sentiment Analysis as POSITIVE, NEGATIVE or NEUTRAL for all the 20 records in the csv file sent \
Also update the urgency and suggested reponse back to the customer for each record. \
Provide the response back in JSON format"

class userresponse(BaseModel):
  record_number :str
  sentiment     :str
  urgency       :str
  reponse_user  :str

#response =  client.models.generate_content( \
  #              model = "gemini-3.6-flash", \
  #              contents = [csvfile, myprompt], \
 #               config-types.GenerateContentConfig( \
 #                   response_mime_type = "application/json", \
 #                   response_schema = list[userresponse], \
 #               ))


response = client.models.generate_content( \
    model="gemini-3.6-flash", \
    contents=[csvfile,myprompt], \
    config=types.GenerateContentConfig( \
        response_mime_type="application/json", \
        response_schema=list[userresponse],  # Pass the Pydantic class here \
    ))


print(response.text)
```
