```python
# Below code is for converting a Python dictionary to JSON string

rishi_dict = {}
rishi_dict[1] = "Roti"
rishi_dict[2] = "Kapda"
rishi_dict[3] = "Makaan"

print ("The value in Rishi dictionary is ",rishi_dict)

import json

rishi_jsonstr = json.dumps(rishi_dict)

print("The value in Rishi JSON converted string is ", rishi_jsonstr)
```

```python
# To convert a json string into a Python dictionary

rishi_jsonstr1 = '{"1" : " Jyoti", "2" : "JanuKunju", "3" : "Rishu"}'
print(rishi_jsonstr1)

import json

rishi_pydict = json.loads(rishi_jsonstr1)
print(rishi_pydict)



```

```python
# To convert a JSON string in the form of an array into a Python list

rishi_jsonstr2 = '["Amar", "Akbar", "Anthony"]'
print(rishi_jsonstr2)

import json

rishi_pylist = json.loads(rishi_jsonstr2)
print(rishi_pylist)

print(rishi_pylist[2])


```

```python
# To convert a json string containing an array of object...to getconverted into a python list of dictionaries

rishi_jsonarrobj = '[{"John":"1", "Johny":"2", "Janardhan":"3" },{"John":"11", "Johny" : "22", "Janardhan" : "33"}]'
print(rishi_jsonarrobj)

import json
rishi_pylist3 = json.loads(rishi_jsonarrobj)
print(rishi_pylist3)
print("The second valueis ", rishi_pylist3[1])
print(" the second value inside the py dict retrieved is ", rishi_pylist3[1].get("Janardhan"))




```

```python
# Code to write a python dictionary to a file in the JSON format

rishi_py_dict = {"1" : "Eena", "2" : "Meena", "3" : "Deeka"}
print(rishi_py_dict)

import json

with open("rishi_file.json","w") as myRishifile:
  json.dump(rishi_py_dict, myRishifile )

print(" The contents of the file has been written")
```

```python
#This code is to open the JSON file and read into the python object

import json

with open("rishi_file.json","r") as rishi_py_file:
  rishi_py_readobj = json.load(rishi_py_file)

  print(rishi_py_readobj)

```

```python
# 2 step process code. one is to write a python list of dictionaries to a JSON file.
# Second step is to read the JSON file and load that into a Python list of dictionaries

rishi_pylistofdict = [\
    {"1" : "GEA", "2" : "Ruia", "3" : "RAIT"},\
    {"1" : "Loretto", "2" : "SIES", "3" : "RAIT1"}\
]
import json
with open("rishi_write_file.json","w") as rishi_pyobj_1:
  json.dump(rishi_pylistofdict, rishi_pyobj_1 )

print("Success")

with open("rishi_write_file.json","r") as rishi_pywriteread:
  rishi_pylistdict = json.load(rishi_pywriteread)

print(rishi_pylistdict)

print("The school of Jyoti is ", rishi_pylistdict[1].get("1") )


```
